```python
from typing import TypedDict, List
from langgraph.graph import StateGraph, START, END
from langchain_core.documents import Document
```

```python
class GraphState(TypedDict):
    question: str
    rewritten_query: str
    documents: List[Document]
    relevant_docs: List[Document]
    generation: str
    retry_count: int
```

```python
def query_analysis(state: GraphState):

    question = state["question"]

    prompt = f"""
    Rewrite the following question for better document retrieval.

    Rules:
    - Return ONLY one improved question
    - Keep the original meaning
    - Do not explain anything
    - Do not give multiple options

    Question:
    {question}
    """

    response = LLM.invoke(prompt)

    return {
        "rewritten_query": response.content.strip()
    }
```

```python
state = {
    "question": "What is runnable pass through?"
}

query_analysis(state)
```

```python
retriever = vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)
```

```python
def retrieve_documents(state: GraphState):

    query = state["rewritten_query"]

    retrieved_docs = retriever.invoke(query)

    return {
        "documents": retrieved_docs
    }
```

```python
state_with_rewritten_query = query_analysis(state)
state.update(state_with_rewritten_query)
docs = retrieve_documents(state)["documents"]

for doc in docs[:2]:
    print(doc.page_content[:300])
    print("=" * 50)
```

```python
def document_grading(state: GraphState):

    question = state["question"]
    documents = state["documents"]

    relevant_docs = []

    for doc in documents:

        prompt = f"""
        You are a document relevance grader.

        Question:
        {question}

        Document:
        {doc.page_content}

        Decide whether the document is relevant to answering
        the question.

        Return only:
        yes
        or
        no
        """

        response = LLM.invoke(prompt)

        grade = response.content.strip().lower()

        if grade == "yes":
            relevant_docs.append(doc)

    return {
        "relevant_docs": relevant_docs
    }
```

```python
state = {
    "question": "What is LangChain?",
    "documents": docs
}

graded_result = document_grading(state)
state.update(graded_result)

state
```

```python
graded_docs = graded_result["relevant_docs"]

print(len(graded_docs))
```

```python
def route_documents(state: GraphState):

    relevant_docs = state["relevant_docs"]
    retry_count = state.get("retry_count", 0)

    # If relevant documents exist
    if len(relevant_docs) > 0:
        return "generate"

    # Retry only 2 times
    elif retry_count < 2:
        return "rewrite"

    # Stop if retry limit reached
    else:
        return "stop"
```

```python
state = {
    "relevant_docs": graded_docs,
    "retry_count": 0
}

route_documents(state)
```

```python
def rewrite_query(state: GraphState):

    question = state["question"]
    retry_count = state.get("retry_count", 0)

    prompt = f"""
    Rewrite this question to improve document retrieval.

    Return only one rewritten question.

    Question:
    {question}
    """

    response = LLM.invoke(prompt)

    return {
        "rewritten_query": response.content.strip(),
        "retry_count": retry_count + 1
    }
```

```python
state = {
    "question": "What is runnable pass through?",
    "retry_count": 0
}

rewrite_query(state)
```

```python
def generate_answer(state: GraphState):

    question = state["question"]
    relevant_docs = state["relevant_docs"]

    context = "\n\n".join(
        [doc.page_content for doc in relevant_docs]
    )

    sources = list(set([
        doc.metadata.get("source", "Unknown")
        for doc in relevant_docs
    ]))

    prompt = f"""
    Answer the question using only the context below.

    Context:
    {context}

    Question:
    {question}

    Give a clear answer.
    """

    response = LLM.invoke(prompt)

    final_answer = response.content

    citation_text = "\n\nSources:\n" + "\n".join(sources)

    return {
        "generation": final_answer + citation_text
    }
```

```python
state = {
    "question": "What is LangChain?",
    "relevant_docs": graded_docs
}

generate_answer(state)
```

```python
def stop_generation(state: GraphState):

    return {
        "generation":
        "I don't know based on available documentation."
    }
```

```python
workflow = StateGraph(GraphState)
```

```python
workflow.add_node("query_analysis", query_analysis)
workflow.add_node("retrieve", retrieve_documents)
workflow.add_node("grade_documents", document_grading)
workflow.add_node("rewrite", rewrite_query)
workflow.add_node("generate", generate_answer)
workflow.add_node("stop", stop_generation)
```

```python
workflow.add_edge(START, "query_analysis")
workflow.add_edge("query_analysis", "retrieve")
workflow.add_edge("retrieve", "grade_documents")
```

```python
workflow.add_conditional_edges(
    "grade_documents",
    route_documents,
    {
        "generate": "generate",
        "rewrite": "rewrite",
        "stop": "stop"
    }
)
```

```python
workflow.add_edge("rewrite", "retrieve")
workflow.add_edge("generate", END)
workflow.add_edge("stop", END)
```

```python
graph = workflow.compile()
```

```python
result = graph.invoke({
    "question": "What is LangChain?",
    "retry_count": 0
})
result["generation"]
```
